```python
import matplotlib.pyplot as plt
```

```python
# Data for 5 Group 14 Elements (C, Si, Ge, Sn, Pb)
atomic_numbers = [6, 14, 32, 50, 82]
bandgaps = [5.47, 1.11, 0.67, 0.1, 0.0]
elements = ['C', 'Si', 'Ge', 'Sn', 'Pb']
```

```python
# Create the plot
plt.figure(figsize=(4, 2))
plt.plot(atomic_numbers, bandgaps, marker='*', linestyle='-', color='g')

# Figure Labelling Basics
plt.title('Bandgap vs. Atomic Number for Group 14 Elements')
plt.xlabel('Atomic Number')
plt.ylabel('Bandgap (eV)')
plt.grid(True)

# Add element labels to the data points
for i, element in enumerate(elements):
    plt.annotate(element, (atomic_numbers[i], bandgaps[i]), xytext=(5,5), textcoords='offset points')

# Display the plot
plt.show()
```
