```python
import pandas as pd
import numpy as np

# Dictionary containing materials data (Students should expand this to 15 entries)
data = {
    'Material': ['Steel', 'Aluminum', 'Copper', 'Alumina', 'Silicon Carbide', 'Nylon', 'Teflon'],
    'Class': ['Metal', 'Metal', 'Metal', 'Ceramic', 'Ceramic', 'Polymer', 'Polymer'],
    'Density_gcm3': [7.85, 2.70, 8.96, 3.95, 3.21, 1.15, 2.20],
    'Bandgap_eV': [0.0, 0.0, 0.0, 8.8, 3.2, 8.5, 9.0],  # Metals have 0 bandgap here for simplicity
    'Thermal_Cond_WmK': [50.2, 237.0, 401.0, 30.0, 120.0, 0.25, 0.25],
    'Tensile_Strength_MPa': [400, 90, 210, 300, 344, 85, 25],
    'Melting_Point_C': [1520, 660, 1085, 2072, 2730, 220, 327],
    'Heat_Capacity_JgK': [0.49, 0.90, 0.38, 0.88, 0.75, 1.70, 1.05]
}

# Convert to DataFrame
df = pd.DataFrame(data)
display(df.head())
```

```python
# 1. Boolean Indexing: Isolate only Ceramics
ceramics_df = df[df['Class'] == 'Ceramic']
print("Ceramics Only:")
display(ceramics_df)

# 2. Complex Filtering: Find materials with High Thermal Conductivity AND high Melting Point
high_temp_conductors = df[(df['Thermal_Cond_WmK'] > 100) & (df['Melting_Point_C'] > 1000)]
print("\nHigh Temperature Conductors:")
display(high_temp_conductors)
```

```python
# Groupby operation: Calculate the mean of numeric columns for each Material Class
# We drop the 'Material' column as it is a non-numeric string identifier
class_means = df.drop(columns=['Material']).groupby('Class').mean()

print("\nMean Properties by Material Class:")
display(class_means[['Bandgap_eV', 'Density_gcm3', 'Thermal_Cond_WmK']])
```
