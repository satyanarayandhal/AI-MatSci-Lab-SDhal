```python
import pandas as pd

# 1. Create a quick dummy materials DataFrame (if they don't have S04 loaded)
data = {
    'Material': ['Fe', 'Al', 'Ti', 'SiC', 'Nylon'],
    'Class': ['Metal', 'Metal', 'Metal', 'Ceramic', 'Polymer'],
    'Melting_Temp_K': [1811.0, 933.47, 1941.0, 3003.0, 538.15]
}
df = pd.DataFrame(data)

# 2. The Core Activity: Unit Conversion (K → °C)
# Subtract 273.15 from the Kelvin column to create a new Celsius column
df['Melting_Temp_C'] = df['Melting_Temp_K'] - 273.15

# 3. Verify the transformation
df.head()
```

```python
# 1. The Core Activity: Categorical Encoding
# Use pd.get_dummies() to convert the 'Class' column into binary indicator columns
# We use dtype=int to ensure the output is 0s and 1s instead of True/False
df_encoded = pd.get_dummies(df, columns=['Class'], dtype=int)

# 2. Verify the transformation
df_encoded.head()
```
